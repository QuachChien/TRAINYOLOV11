```python
from ultralytics import YOLO

data_yaml = "/content/drive/MyDrive/test_roboflow/data.yaml"

# 1. Sử dụng mô hình Segmentation (yolo11n-seg.pt) thay vì mô hình Detection
model = YOLO("yolo11n-seg.pt")

# 2. Tiến hành huấn luyện Instance Segmentation
results = model.train(
    data=data_yaml,
    epochs=100,
    imgsz=640,
    batch=64,           # Có thể giảm xuống 32 nếu bị tràn bộ nhớ GPU (Out Of Memory)
    patience=50,
    cache=True,
    workers=8,
    name="test_seg_20261018",
    exist_ok=True,
    amp=True
)

# 3. Sao lưu tự động toàn bộ kết quả Segmentation vào Google Drive
!mkdir -p /content/drive/MyDrive/yolo11_backup/
!cp -r /content/runs/segment/test_seg_20261018 /content/drive/MyDrive/yolo11_backup/
print("✅ Đã backup thành công kết quả SEGMENTATION vào thư mục 'yolo11_backup' trên Google Drive!")
```

```python
# Tiến hành cài đặt thư viện thiếu để sửa lỗi ModuleNotFoundError
%pip install onnxscript

from ultralytics import YOLO
import os

# Đường dẫn tới file trọng số tốt nhất (.pt) sau khi train Segmentation xong
model_path = "/content/runs/segment/test_seg_20261018/weights/best.pt"

if os.path.exists(model_path):
    # Tải mô hình
    model = YOLO(model_path)

    # Tiến hành xuất (export) sang định dạng ONNX
    # Bạn có thể thêm opset=12 nếu cần tương thích với một số hệ thống cũ
    onnx_path = model.export(format="onnx", imgsz=640, dynamic=False)
    print(f"\n✅ Đã chuyển đổi thành công sang ONNX! File lưu tại: {onnx_path}")

    # Sao lưu file ONNX sang Google Drive
    !cp {onnx_path} /content/drive/MyDrive/yolo11_backup/
    print("✅ Đã sao lưu file ONNX sang thư mục 'yolo11_backup' trên Google Drive!")
else:
    print(f"❌ Không tìm thấy file trọng số tại {model_path}. Vui lòng kiểm tra lại quá trình train đã hoàn tất chưa.")
```
